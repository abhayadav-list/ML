## views vs copies

view me any change to the view arry will also change the original arry
view=arr
now if we do 
view[0]=22
then 
arr[0] will also become 22

but in case of copy it will not reflect the changes as copy creates a total new arr

arrc=np.copy(arr)
arrc[0]=22
will not affect the original arr

arr[0]==original value


```python
#copy
arr= np.random.randint(0,10,size=(2,3))
arrc=np.copy(arr)
arrc[0][0]=22
print(arr)
print(arrc)
```

```python
#view 
arr= np.random.randint(0,10,size=(2,3))
arrc=arr
arrc[0][0]=22
print(arr)
print(arrc)
```

## Concat
np.concatenate
# Note :
ValueError: all the input array dimensions except for the concatenation axis must match exactly

```python
a=np.random.randint(1,10,size=(2,2))
b=np.random.randint(2,10,size=(2,2))
combined=np.concatenate((a,b))
print(a)
print(b)
print(combined)
print(combined.shape)
```

```python
combine=np.hstack((a,b)) #concate hroizontally 
print(combine)
print(combine.shape)
```

```python
combine1=np.vstack((a,b)) #concate vertically 
print(combine1)
print(combine1.shape)
```

```python
# this can be also done with the above concatenate function as well instead of using the 
# hstack or the vstack 

combine=np.concatenate((a,b),0) # here we are also giving the axes for concat , 0->horizontal and 1-> vertical axes
print(combine)
print(combine.shape)
```

```python
combine_1=np.concatenate((a,b),1)
print(combine_1)
print(combine_1.shape)
```
